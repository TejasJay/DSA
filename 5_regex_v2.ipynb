# 📘 Topic 4: Assertions (Lookahead/Lookbehind)

## 1️⃣ Topic Overview (Theory)

**What are Assertions?**

Assertions (also called lookahead and lookbehind) are zero-width patterns that check if a condition is true at a specific position without consuming any characters. They "look ahead" or "look behind" without moving the match position.

**Why do they exist?**

They allow you to match patterns based on what comes before or after, without including that context in the match itself. This is powerful for complex matching scenarios.

**How Python `re` interprets them:**

- Python's `re` module supports lookahead and lookbehind assertions
- They are zero-width (don't consume characters)
- Lookbehind has limitations: fixed-width patterns only in some cases

**Common mistakes:**

- Forgetting that assertions don't consume characters
- Using variable-width patterns in lookbehind (not always supported)
- Confusing positive and negative assertions

**Mental model:** Think of assertions as "peeking" ahead or behind without moving forward—like checking if a door is locked before trying to open it.


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `?=` (Positive lookahead assertion)

**Explanation:**
- Matches a pattern only if it's followed by another pattern
- Does NOT consume any characters
- Syntax: `pattern(?=lookahead)`

**What problem it solves:** Matching a pattern only when it's followed by a specific sequence.

**When NOT to use it:** When you want to include the following pattern in the match (use a regular group instead).


```python
# Problem: Find "Python" only when followed by "3" (but don't include "3" in match)
text = "Python3 is great, but Python2 is old"

# Solution: Use positive lookahead (?=)
# Python(?=3) matches "Python" only if followed by "3"
# The "3" is not included in the match
pattern = r'Python(?=3)'

matches = re.findall(pattern, text)
print(f"Matches: {matches}")
print("✓ Matched 'Python' only when followed by '3', but '3' not included")

# Compare with regular match (includes the 3)
regular_match = re.findall(r'Python3', text)
print(f"Regular match (includes 3): {regular_match}")
```

### 🔹 Sub-topic: `?!` (Negative lookahead assertion)

**Explanation:**
- Matches a pattern only if it's NOT followed by another pattern
- Does NOT consume any characters
- Syntax: `pattern(?!lookahead)`

**What problem it solves:** Matching a pattern only when it's NOT followed by a specific sequence.

**When NOT to use it:** When you want to match when it IS followed by something (use positive lookahead).


```python
# Problem: Find "Python" only when NOT followed by "3"
text = "Python3 is great, but Python2 is old, and Python is awesome"

# Solution: Use negative lookahead (?!)
# Python(?!3) matches "Python" only if NOT followed by "3"
pattern = r'Python(?!3)'

matches = re.findall(pattern, text)
print(f"Matches: {matches}")
print("✓ Matched 'Python' only when NOT followed by '3'")
```

### 🔹 Sub-topic: `?<=` (Positive lookbehind assertion)

**Explanation:**
- Matches a pattern only if it's preceded by another pattern
- Does NOT consume any characters
- Syntax: `(?<=lookbehind)pattern`
- **⚠️ Limitation:** Lookbehind must be fixed-width in Python's `re` module

**What problem it solves:** Matching a pattern only when it's preceded by a specific sequence.

**When NOT to use it:** When you want to include the preceding pattern in the match.


```python
# Problem: Extract price value (number) only when preceded by "$"
text = "Price: $99.99, Cost: €50.00"

# Solution: Use positive lookbehind (?<=)
# (?<=\$) matches position after "$"
# \\d+\\.\\d+ matches the price number
# The "$" is not included in the match
pattern = r'(?<=\$)\d+\.\d+'

matches = re.findall(pattern, text)
print(f"Price values after $: {matches}")
print("✓ Matched price only when preceded by '$', but '$' not included")
```

### 🔹 Sub-topic: `?<!` or `?<-` (Negative lookbehind assertion)

**Explanation:**
- Matches a pattern only if it's NOT preceded by another pattern
- Does NOT consume any characters
- Syntax: `(?<!lookbehind)pattern` or `(?<-lookbehind)pattern`
- **⚠️ Limitation:** Lookbehind must be fixed-width in Python's `re` module

**What problem it solves:** Matching a pattern only when it's NOT preceded by a specific sequence.

**When NOT to use it:** When you want to match when it IS preceded by something.


```python
# Problem: Find numbers NOT preceded by "$"
text = "Price: $99.99, Cost: 50.00, Total: 149.99"

# Solution: Use negative lookbehind (?<!)
# (?<!\\$) matches position NOT after "$"
# \\d+\\.\\d+ matches the number
pattern = r'(?<!\$)\d+\.\d+'

matches = re.findall(pattern, text)
print(f"Numbers not after $: {matches}")
print("✓ Matched numbers only when NOT preceded by '$'")
```

### 🔹 Sub-topic: `?>` (Once-only subexpression)

**Explanation:**
- Prevents backtracking into the subexpression
- **⚠️ Not supported in Python's `re` module**
- This is an atomic group feature from other regex engines

**What problem it solves:** Performance optimization by preventing backtracking (advanced feature).

**Python alternative:** Python's `re` doesn't support atomic groups, but you can structure patterns to minimize backtracking.

**Explanation:**
- Conditional pattern matching: if a group matched, then match one pattern, else match another
- **⚠️ Not supported in Python's `re` module**
- Syntax varies by regex engine

**What problem it solves:** Pattern matching based on whether a previous group matched.

**Python alternative:** Use Python's conditional logic outside regex or restructure the pattern.

### 🔹 Sub-topic: `?#` (Comment)

**Explanation:**
- Allows comments within regex patterns
- **⚠️ Requires `re.VERBOSE` flag in Python**
- Syntax: `(?#comment text)`

**What problem it solves:** Documenting complex regex patterns inline.

**When to use it:** For complex patterns that need explanation.


```python
# Problem: Use comments in regex pattern for documentation
text = "Email: user@example.com"

# Solution: Use (?#comment) with re.VERBOSE flag
# re.VERBOSE allows whitespace and comments in patterns
pattern = r'''
    \w+           # Username part (word characters)
    @             # Literal @ symbol
    \w+           # Domain name (word characters)
    \.             # Literal dot
    \w+           # Top-level domain (word characters)
'''

# Use re.VERBOSE to enable comments and ignore whitespace
match = re.search(pattern, text, re.VERBOSE)
if match:
    print(f"Found email: {match.group()}")
    print("✓ Comments help document complex patterns")
```

# 📘 Topic 5: Quantifiers

## 1️⃣ Topic Overview (Theory)

**What are Quantifiers?**

Quantifiers specify how many times a character, group, or character class should be matched. They control the "quantity" or repetition of patterns.

**Why do they exist?**

They allow you to match variable-length patterns without specifying each repetition explicitly. Essential for matching "one or more", "zero or more", "exactly N times", etc.

**How Python `re` interprets them:**

- Quantifiers work the same in Python's `re` module
- By default, quantifiers are "greedy" (match as much as possible)
- Add `?` after a quantifier to make it "lazy" or "non-greedy" (match as little as possible)

**Common mistakes:**

- Forgetting that `*` means "zero or more" (can match nothing!)
- Not understanding greedy vs lazy behavior
- Using `+` when you mean `*` or vice versa

**Mental model:** Quantifiers are "multipliers" - they tell the regex engine "match this pattern X times" where X can be a range.


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `*` (0 or more)

**Explanation:**
- Matches zero or more occurrences of the preceding element
- Greedy by default (matches as much as possible)
- Example: `a*` matches "", "a", "aa", "aaa", etc.

**What problem it solves:** Matching optional or repeated patterns.

**When NOT to use it:** When you need at least one match (use `+` instead).


```python
# Problem: Match optional "www." prefix in URLs
text = "https://example.com, https://www.example.com, https://test.com"

# Solution: Use * for zero or more
# (www\\.)? is equivalent, but * shows the concept
# www\\. matches "www."
# * means zero or more (so it's optional)
pattern = r'https://(www\.)*\w+\.\w+'

matches = re.findall(pattern, text)
print(f"URLs found: {matches}")
print("✓ * allows zero or more matches (optional pattern)")

# Note: * on its own matches the empty string
empty_match = re.findall(r'a*', 'b')
print(f"a* on 'b' matches: {empty_match} (empty string)")
```

### 🔹 Sub-topic: `+` (1 or more)

**Explanation:**
- Matches one or more occurrences of the preceding element
- Greedy by default
- Example: `a+` matches "a", "aa", "aaa", but NOT ""

**What problem it solves:** Matching patterns that must appear at least once.

**When NOT to use it:** When the pattern is optional (use `*` or `?`).


```python
# Problem: Extract words (sequences of letters, at least one)
text = "Hello world! 123 test"

# Solution: Use + for one or more
# \\w+ matches one or more word characters
# This ensures we get actual words, not empty matches
pattern = r'\w+'

matches = re.findall(pattern, text)
print(f"Words found: {matches}")
print("✓ + requires at least one match (no empty strings)")

# Compare with * (would also match empty strings between non-word chars)
star_matches = re.findall(r'\w*', text)
print(f"\\w* matches: {star_matches} (includes empty strings)")
```

### 🔹 Sub-topic: `?` (0 or 1)

**Explanation:**
- Matches zero or one occurrence (makes the preceding element optional)
- Greedy by default
- Example: `a?` matches "" or "a"

**What problem it solves:** Making patterns optional.

**When NOT to use it:** When you need multiple occurrences (use `+` or `*`).


```python
# Problem: Match phone numbers with optional country code
text = "Call +1-555-1234 or 555-5678"

# Solution: Use ? for optional pattern
# (\\+\\d+-)? makes the country code optional
# \\+\\d+- matches "+1-" format
# ? means zero or one (optional)
pattern = r'(\+\d+-)?\d{3}-\d{4}'

matches = re.findall(pattern, text)
print(f"Phone numbers: {matches}")
print("✓ ? makes patterns optional (zero or one occurrence)")

# Extract full matches
full_matches = re.findall(r'(\+\d+-)?\d{3}-\d{4}', text)
for match in full_matches:
    phone = ''.join(match) if match[0] else ''
    remaining = re.search(r'(\+\d+-)?(\d{3}-\d{4})', text)
    if remaining:
        print(f"Found: {remaining.group()}")
        text = text.replace(remaining.group(), '', 1)
```

### 🔹 Sub-topic: `{n}` (Exactly n)

**Explanation:**
- Matches exactly n occurrences of the preceding element
- Example: `a{3}` matches "aaa" only

**What problem it solves:** Matching a specific number of repetitions.

**When NOT to use it:** When you need a range (use `{n,m}`).


```python
# Problem: Match exactly 4-digit years
text = "Born in 1998, graduated in 2024, year 99 is invalid"

# Solution: Use {n} for exact count
# \\d{4} matches exactly 4 digits
pattern = r'\d{4}'

matches = re.findall(pattern, text)
print(f"4-digit years: {matches}")
print("✓ {n} matches exactly n occurrences")

# Test with different lengths
test_text = "123 1234 12345"
exact_4 = re.findall(r'\d{4}', test_text)
print(f"Exactly 4 digits from '{test_text}': {exact_4}")
```

### 🔹 Sub-topic: `{n,}` (n or more)

**Explanation:**
- Matches n or more occurrences
- Example: `a{3,}` matches "aaa", "aaaa", "aaaaa", etc.

**What problem it solves:** Matching at least n occurrences.

**When NOT to use it:** When you need an upper limit (use `{n,m}`).


```python
# Problem: Match phone numbers with at least 10 digits
text = "Call 1234567890 or 123-456-7890 or 12345"

# Solution: Use {n,} for n or more
# \\d{10,} matches 10 or more digits
pattern = r'\d{10,}'

matches = re.findall(pattern, text)
print(f"Phone numbers (10+ digits): {matches}")
print("✓ {n,} matches n or more occurrences")

# Test minimum requirement
test_cases = ["123456789", "1234567890", "12345678901"]
for test in test_cases:
    if re.search(pattern, test):
        print(f"  '{test}' matches (10+ digits)")
    else:
        print(f"  '{test}' doesn't match (< 10 digits)")
```

### 🔹 Sub-topic: `{n,m}` (Between n and m)

**Explanation:**
- Matches between n and m occurrences (inclusive)
- Example: `a{2,4}` matches "aa", "aaa", "aaaa"

**What problem it solves:** Matching a range of repetitions.

**When NOT to use it:** When you need exact count (use `{n}`) or unlimited (use `+` or `*`).

### 🔹 Sub-topic: Lazy/Non-greedy quantifiers (`*?`, `+?`, `??`, `{n,m}?`)

**Explanation:**
- Adding `?` after a quantifier makes it "lazy" or "non-greedy"
- Lazy quantifiers match as little as possible (opposite of greedy)
- Examples: `.*?` (lazy), `.+?` (lazy), `.{2,5}?` (lazy)

**What problem it solves:** Matching the shortest possible sequence instead of the longest.

**When NOT to use it:** When you want the longest match (use greedy quantifiers).


```python
# Problem: Extract content between first set of quotes (shortest match)
text = 'Say "hello" and "goodbye" to friends'

# Greedy match (default) - matches longest possible
greedy_pattern = r'"(.*)"'
greedy_match = re.search(greedy_pattern, text)
if greedy_match:
    print(f"Greedy match: '{greedy_match.group(1)}'")
    print("  (matched from first \" to last \")")

# Lazy match - matches shortest possible
lazy_pattern = r'"(.*?)"'
lazy_matches = re.findall(lazy_pattern, text)
print(f"Lazy matches: {lazy_matches}")
print("  (matched each quoted string separately)")
print("✓ Lazy quantifiers (*?, +?) match as little as possible")
```

# 📘 Topic 6: Escape Sequences

## 1️⃣ Topic Overview (Theory)

**What are Escape Sequences?**

Escape sequences allow you to match special characters literally or use special character codes. They use a backslash (`\`) to "escape" the special meaning of characters or to represent non-printable characters.

**Why do they exist?**

Many characters have special meanings in regex (like `.`, `*`, `+`). To match them literally, you need to escape them. Also, some characters like newlines can't be typed directly, so escape sequences represent them.

**How Python `re` interprets them:**

- Python's `re` module uses backslash for escaping
- In raw strings (`r"..."`), backslashes are treated literally by Python, then by the regex engine
- Always use raw strings for regex patterns in Python

**Common mistakes:**

- Not using raw strings (causing double-escaping issues)
- Forgetting to escape special characters that should be literal
- Confusing Python string escapes with regex escapes

**Mental model:** Escape sequences are "translators" - they tell the regex engine "treat this character literally" or "this represents a special character."


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `\` (Escape following character)

**Explanation:**
- Backslash escapes the following character
- Makes special characters literal, or gives special meaning to ordinary characters
- Example: `\.` matches a literal dot, `\\` matches a literal backslash

**What problem it solves:** Matching special regex characters as literal characters.

**When NOT to use it:** When you want the special meaning of the character (don't escape it).


```python
# Problem: Match a literal dot in a filename
filename = "document.pdf"

# Solution: Escape the dot with backslash
# \\. matches a literal dot (not "any character")
# Without escape, . would match any character
pattern = r'\.pdf$'

if re.search(pattern, filename):
    print(f"✓ {filename} ends with .pdf")
else:
    print("✗ Not a PDF file")

# Escape other special characters
special_text = "Price: $99.99 (50% off)"
# Match literal $, ., (, ), %
pattern_special = r'\$99\.99 \(50% off\)'
if re.search(pattern_special, special_text):
    print("✓ Matched literal special characters")
print("✓ Backslash escapes special characters to match them literally")
```

### 🔹 Sub-topic: `\Q` and `\E` (Begin/End literal sequence)

**Explanation:**
- `\Q` starts a literal sequence (treats all characters literally)
- `\E` ends the literal sequence
- **⚠️ Not directly supported in Python's `re` module**
- Python alternative: Use `re.escape()` function

**What problem it solves:** Escaping multiple special characters at once.

**Python alternative:** Use `re.escape()` to escape all special characters in a string.


```python
# Problem: Match a string with many special characters
# Note: \\Q and \\E are not standard in Python re
# We use re.escape() instead

special_string = "Price: $99.99 (50% off!)"

# Solution: Use re.escape() to escape all special characters
# re.escape() is Python's equivalent of \\Q...\\E
escaped = re.escape("$99.99 (50% off!)")
pattern = r'Price: ' + escaped

if re.search(pattern, special_string):
    print("✓ Matched escaped special characters")
    print(f"Escaped pattern: {pattern}")
print("✓ In Python, use re.escape() instead of \\Q...\\E")
```

**What are Special Characters?**

Special characters in regex represent non-printable characters like newlines, tabs, and carriage returns. They use escape sequences to represent characters that can't be typed directly.

**Why do they exist?**

Text often contains invisible characters (newlines, tabs) that need to be matched. Special character escapes provide a way to match these characters in patterns.

**How Python `re` interprets them:**

- Python's `re` module supports standard escape sequences
- These work the same as in Python strings
- Always use raw strings to avoid double-escaping issues

**Common mistakes:**

- Not using raw strings (causing escape sequence issues)
- Confusing regex escapes with Python string escapes

**Mental model:** Special characters are "invisible characters" - they exist in text but aren't visible when printed.


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `\n` (New line)

**Explanation:**
- Matches a newline character (line feed, ASCII 10)
- In Python, this is the character that separates lines

**What problem it solves:** Matching or splitting text at line boundaries.

**When NOT to use it:** When you want to match any whitespace (use `\s`).


```python
# Problem: Split text into lines
text = "Line 1\nLine 2\nLine 3"

# Solution: Use \\n to match newlines
# Split on newline character
lines = re.split(r'\n', text)
print(f"Lines: {lines}")
print(f"Number of lines: {len(lines)}")

# Find all newlines
newlines = re.findall(r'\n', text)
print(f"Newline characters found: {len(newlines)}")
print("✓ \\n matches newline characters")
```

**Explanation:**
- Matches a carriage return character (ASCII 13)

```python
# Problem: Handle Windows line endings (\\r\\n)
text = "Line 1\r\nLine 2\r\nLine 3"

# Solution: Match carriage return
# \\r matches carriage return
# Windows uses \\r\\n for line endings
pattern = r'\r\n'

lines = re.split(pattern, text)
print(f"Lines (Windows format): {lines}")

# Match just carriage return
carriage_returns = re.findall(r'\r', text)
print(f"Carriage returns found: {len(carriage_returns)}")
print("✓ \\r matches carriage return characters")
```

# 📘 Topic 8: Groups and Ranges

## 1️⃣ Topic Overview (Theory)

**What are Groups and Ranges?**

Groups allow you to capture parts of a match and apply quantifiers to multiple characters. Ranges specify character sets using character classes.

**Why do they exist?**

Groups enable you to extract specific parts of matches, apply operations to multiple characters, and create reusable subpatterns. Ranges provide concise ways to match character sets.

**How Python `re` interprets them:**

- Groups are numbered starting from 1 (0 is the entire match)
- Capturing groups `(...)` store matched content
- Non-capturing groups `(?:...)` don't store matches
- Ranges work within character classes `[a-z]`

**Common mistakes:**

- Forgetting that groups are numbered from 1, not 0
- Not understanding the difference between capturing and non-capturing groups
- Confusing alternation `|` with character classes

**Mental model:** Groups are "containers" that capture and organize parts of matches. Ranges are "shortcuts" for character sets.


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `.` (Any character except new line)

**Explanation:**
- Matches any single character except newline (`\n`)
- In DOTALL mode (`re.DOTALL`), also matches newline
- Very commonly used

**What problem it solves:** Matching any character when you don't care what it is.

**When NOT to use it:** When you need to match newlines too (use `[\s\S]` or enable DOTALL).


```python
# Problem: Match pattern with any character in between
text = "a1b aXb a b a9b"

# Solution: Use . to match any character
# . matches any character except newline
pattern = r'a.b'

matches = re.findall(pattern, text)
print(f"Matches: {matches}")
print("✓ . matches any single character (except newline)")

# Test with newline
text_with_newline = "a\nb"
match_newline = re.search(r'a.b', text_with_newline)
if match_newline:
    print("Matched with newline")
else:
    print("✗ . does not match newline by default")

# With DOTALL flag, . matches newline too
match_dotall = re.search(r'a.b', text_with_newline, re.DOTALL)
if match_dotall:
    print("✓ With DOTALL, . matches newline")
```

### 🔹 Sub-topic: `(a|b)` (Alternation - a or b)

**Explanation:**
- Matches either pattern `a` OR pattern `b`
- Can be used with multiple alternatives: `(a|b|c)`
- Groups the alternatives together

**What problem it solves:** Matching one of several possible patterns.

**When NOT to use it:** When you want to match any character from a set (use character class `[abc]` instead).


```python
# Problem: Match "cat" or "dog"
text = "I have a cat and a dog"

# Solution: Use | for alternation
# (cat|dog) matches either "cat" or "dog"
pattern = r'(cat|dog)'

matches = re.findall(pattern, text)
print(f"Matches: {matches}")
print("✓ (a|b) matches either pattern a or b")

# Multiple alternatives
colors = "red, blue, green, yellow"
color_pattern = r'(red|blue|green)'
color_matches = re.findall(color_pattern, colors)
print(f"Colors found: {color_matches}")
```

### 🔹 Sub-topic: `(...)` (Capturing group)

**Explanation:**
- Captures the matched content for later use
- Groups are numbered from 1 (group 0 is entire match)
- Can be referenced with `\1`, `\2`, etc. (backreferences)
- Can be accessed via `match.group(1)`, `match.group(2)`, etc.

**What problem it solves:** Extracting specific parts of a match.

**When NOT to use it:** When you don't need to capture (use non-capturing group `(?:...)`).


```python
# Problem: Extract area code from phone number
text = "(416)-555-1234"

# Solution: Use capturing group (...)
# (\\d{3}) captures the area code
pattern = r'\((\d{3})\)'

match = re.search(pattern, text)
if match:
    print(f"Full match: {match.group(0)}")  # Group 0 is entire match
    print(f"Area code (group 1): {match.group(1)}")  # Group 1 is first capture
    print("✓ (...) captures matched content")

# Multiple capturing groups
email = "user@example.com"
email_pattern = r'(\w+)@(\w+\.\w+)'
email_match = re.search(email_pattern, email)
if email_match:
    print(f"\\nEmail parts:")
    print(f"  Username: {email_match.group(1)}")
    print(f"  Domain: {email_match.group(2)}")
```

### 🔹 Sub-topic: `(?:...)` (Non-capturing group)

**Explanation:**
- Groups patterns together but doesn't capture the match
- Useful for applying quantifiers to multiple characters
- Doesn't create a numbered group

**What problem it solves:** Grouping without capturing (better performance, cleaner groups).

**When NOT to use it:** When you need to extract the matched content (use capturing group).


```python
# Problem: Match "www." optionally, but don't capture it
text = "https://www.example.com and https://example.com"

# Solution: Use non-capturing group (?:...)
# (?:www\\.)? groups "www." and makes it optional, but doesn't capture
pattern = r'https://(?:www\.)?\w+\.\w+'

matches = re.findall(pattern, text)
print(f"URLs: {matches}")

# Compare: capturing group creates a group
capturing_pattern = r'https://(www\.)?\w+\.\w+'
capturing_matches = re.findall(capturing_pattern, text)
print(f"With capturing group: {capturing_matches}")
print("✓ (?:...) groups without capturing")
```

### 🔹 Sub-topic: `[abc]` (Character class - a or b or c)

**Explanation:**
- Matches any one character from the set
- Equivalent to `(a|b|c)` but more efficient for single characters
- Characters are treated literally (no special meaning)

**What problem it solves:** Matching any character from a specific set.

**When NOT to use it:** When you want to match sequences (use alternation `(a|b)`).


```python
# Problem: Find words starting with 'a', 'b', or 'c'
text = "apple banana cherry dog elephant"

# Solution: Use [abc] character class
# [abc] matches any one character: 'a', 'b', or 'c'
pattern = r'\b[abc]\w+'

matches = re.findall(pattern, text)
print(f"Words starting with a, b, or c: {matches}")
print("✓ [abc] matches any character from the set")

# Compare with alternation (less efficient for single chars)
alt_pattern = r'\b(a|b|c)\w+'
alt_matches = re.findall(alt_pattern, text)
print(f"With alternation: {alt_matches}")
```

### 🔹 Sub-topic: `[^abc]` (Negated character class)

**Explanation:**
- Matches any character NOT in the set
- `^` at the start of character class negates it
- Example: `[^abc]` matches any character except 'a', 'b', or 'c'

**What problem it solves:** Matching characters that are NOT in a specific set.

**When NOT to use it:** When you want to match characters in a set (use `[abc]`).


```python
# Problem: Find words that don't start with vowels
text = "apple banana cherry dog elephant"

# Solution: Use [^aeiou] negated character class
# [^aeiou] matches any character that is NOT a vowel
pattern = r'\b[^aeiou]\w+'

matches = re.findall(pattern, text)
print(f"Words not starting with vowels: {matches}")
print("✓ [^abc] matches any character NOT in the set")
```

### 🔹 Sub-topic: `[a-q]`, `[A-Q]`, `[0-7]` (Character ranges)

**Explanation:**
- Matches any character in the specified range
- `[a-q]` matches lowercase letters from 'a' to 'q'
- `[A-Q]` matches uppercase letters from 'A' to 'Q'
- `[0-7]` matches digits from '0' to '7'
- Ranges are inclusive

**What problem it solves:** Matching characters in a continuous range.

**When NOT to use it:** When you need non-contiguous characters (list them explicitly).


```python
# Problem: Extract different character types using ranges
text = "Hello123World456"

# Solution: Use character ranges
# [a-z] matches lowercase letters
lowercase = re.findall(r'[a-z]+', text)
print(f"Lowercase: {lowercase}")

# [A-Z] matches uppercase letters
uppercase = re.findall(r'[A-Z]+', text)
print(f"Uppercase: {uppercase}")

# [0-9] matches digits (equivalent to \\d)
digits = re.findall(r'[0-9]+', text)
print(f"Digits: {digits}")

# [0-7] matches octal digits only
octal_text = "0123456789"
octal_digits = re.findall(r'[0-7]+', octal_text)
print(f"Octal digits (0-7): {octal_digits}")
print("✓ Character ranges match characters in a continuous range")
```

### 🔹 Sub-topic: `\x` (Backreference to group number x)

**Explanation:**
- References a previously captured group
- `\1` references group 1, `\2` references group 2, etc.
- Matches the same text that was captured by the group

**What problem it solves:** Matching repeated patterns (like duplicate words).

**When NOT to use it:** When you want to match different text (don't use backreference).


```python
# Problem: Find consecutive duplicate words
text = "this is is a test test"

# Solution: Use backreference \\1
# \\b(\\w+)\\s\\1\\b matches a word followed by the same word
# (\\w+) captures the word (group 1)
# \\s matches whitespace
# \\1 matches the same word as group 1
pattern = r'\b(\w+)\s\1\b'

matches = re.findall(pattern, text)
print(f"Duplicate words: {matches}")
print("✓ \\1 references the first capturing group")

# Find repeated patterns
html_tags = "<div>content</div><p>text</p>"
tag_pattern = r'<(\w+)>.*?</\1>'
tag_matches = re.findall(tag_pattern, html_tags)
print(f"\\nMatching tags: {tag_matches}")
print("✓ Backreferences match the same content as the captured group")
```

### Combined Example 1: Using Groups and Ranges Together


**Problem:** Parse a formatted string extracting multiple parts using capturing groups and character ranges


```python
# Problem: Parse product information with multiple fields
# Format: "SKU: ABC-123, Price: $99.99, Stock: 45"
text = "SKU: ABC-123, Price: $99.99, Stock: 45"

# Solution using groups and ranges:
# ([A-Z]+) captures uppercase letters (SKU prefix)
# (\\d+) captures digits (SKU number, price, stock)
# Character ranges [A-Z] and [0-9] for specific types

# Extract SKU using capturing groups and ranges
sku_pattern = r'SKU: ([A-Z]+)-(\d+)'
sku_match = re.search(sku_pattern, text)
if sku_match:
    print(f"SKU prefix: {sku_match.group(1)}")
    print(f"SKU number: {sku_match.group(2)}")

# Extract price
price_pattern = r'Price: \\$(\d+\.\d+)'
price_match = re.search(price_pattern, text)
if price_match:
    print(f"Price: ${price_match.group(1)}")

# Extract stock
stock_pattern = r'Stock: (\d+)'
stock_match = re.search(stock_pattern, text)
if stock_match:
    print(f"Stock: {stock_match.group(1)}")

print("\\n✓ Combined groups and ranges for complex parsing")
```

### Combined Example 2: Backreferences and Non-capturing Groups


**Problem:** Find HTML-like tags with matching opening and closing tags using backreferences


```python
# Problem: Match HTML-like tags with matching names
html = "<div>content</div><p>text</p><div>more</span>"

# Solution using backreferences:
# <(\\w+)> captures the tag name (group 1)
# .*? matches content (lazy)
# </\\1> matches closing tag with same name (backreference to group 1)
pattern = r'<(\w+)>.*?</\1>'

matches = re.findall(pattern, html)
print(f"Matching tags found: {matches}")

# Extract full tag pairs
full_matches = re.finditer(pattern, html)
for match in full_matches:
    print(f"  Tag: {match.group(1)}, Full: {match.group(0)}")

# Using non-capturing group for optional attributes
tag_with_attr = '<div class="test">content</div>'
attr_pattern = r'<(\w+)(?:\s[^>]*)?>.*?</\1>'
attr_match = re.search(attr_pattern, tag_with_attr)
if attr_match:
    print(f"\\nTag with optional attributes: {attr_match.group(1)}")
print("✓ Backreferences ensure matching opening/closing tags")
```

# 📘 Topic 9: Pattern Modifiers

## 1️⃣ Topic Overview (Theory)

**What are Pattern Modifiers?**

Pattern modifiers (also called flags) change how the regex engine interprets and executes the pattern. They affect behavior like case sensitivity, multiline matching, and dotall mode.

**Why do they exist?**

They provide control over regex behavior without changing the pattern itself. Essential for handling different text formats and matching requirements.

**How Python `re` interprets them:**

- Python's `re` module uses flags passed as arguments or inline flags
- Flags can be combined using `|` (bitwise OR)
- Some flags can be set inline in the pattern using `(?flag)` syntax

**Common mistakes:**

- Forgetting that flags affect the entire pattern
- Not understanding the difference between `re.MULTILINE` and `re.DOTALL`
- Confusing global matching (Python doesn't have a global flag like JavaScript)

**Mental model:** Modifiers are "settings" that change how the regex engine works, not what it matches.


## 2️⃣ Sub-topic Walkthrough

### 🔹 Sub-topic: `i` (Case-insensitive)

**Explanation:**
- Makes the pattern case-insensitive
- In Python: `re.IGNORECASE` or `re.I`
- Matches both uppercase and lowercase letters

**What problem it solves:** Matching text regardless of case.

**When NOT to use it:** When case matters (don't use the flag).


```python
# Problem: Match "python" case-insensitively
text = "Python is great, PYTHON is powerful, python is simple"

# Solution: Use re.IGNORECASE flag (or re.I)
# Without flag: only matches exact case
pattern = r'python'
matches_no_flag = re.findall(pattern, text)
print(f"Without flag: {matches_no_flag}")

# With flag: matches all cases
matches_with_flag = re.findall(pattern, text, re.IGNORECASE)
print(f"With IGNORECASE flag: {matches_with_flag}")
print("✓ re.IGNORECASE makes pattern case-insensitive")
```

### 🔹 Sub-topic: `m` (Multiple lines)

**Explanation:**
- Makes `^` and `$` match at the start/end of each line, not just the string
- In Python: `re.MULTILINE` or `re.M`
- Each line is treated separately

**What problem it solves:** Matching patterns at the start or end of lines in multiline text.

**When NOT to use it:** When you want `^` and `$` to match only string boundaries.


```python
# Problem: Match lines that start with "ERROR:"
text = """INFO: System started
ERROR: Connection failed
INFO: Process completed
ERROR: Timeout occurred"""

# Without MULTILINE: ^ only matches start of entire string
pattern = r'^ERROR:'
matches_no_flag = re.findall(pattern, text)
print(f"Without MULTILINE: {matches_no_flag}")

# With MULTILINE: ^ matches start of each line
matches_with_flag = re.findall(pattern, text, re.MULTILINE)
print(f"With MULTILINE flag: {matches_with_flag}")
print("✓ re.MULTILINE makes ^ and $ match per line")
```

### 🔹 Sub-topic: `s` (Treat string as single line / DOTALL)

**Explanation:**
- Makes `.` (dot) match newline characters too
- In Python: `re.DOTALL` or `re.S`
- "Single line" means dot matches everything, including newlines

**What problem it solves:** Matching patterns that span multiple lines.

**When NOT to use it:** When you want `.` to stop at newlines (default behavior).


```python
# Problem: Match content between tags that spans multiple lines
html = """<div>
Content here
More content
</div>"""

# Without DOTALL: . does not match newline
pattern = r'<div>(.*)</div>'
match_no_flag = re.search(pattern, html)
if match_no_flag:
    print(f"Without DOTALL: {match_no_flag.group(1)}")
else:
    print("Without DOTALL: No match (newline stops .)")

# With DOTALL: . matches newline too
match_with_flag = re.search(pattern, html, re.DOTALL)
if match_with_flag:
    print(f"With DOTALL: {match_with_flag.group(1).strip()}")
print("✓ re.DOTALL makes . match newline characters")
```

### 🔹 Sub-topic: `x` (Allow comments and whitespace)

**Explanation:**
- Allows whitespace and comments in patterns for readability
- In Python: `re.VERBOSE` or `re.X`
- Comments use `#` and extend to end of line

**What problem it solves:** Making complex regex patterns more readable.

**When NOT to use it:** For simple patterns (adds unnecessary complexity).


```python
# Problem: Create a readable email pattern with comments
text = "Contact: user@example.com"

# Solution: Use re.VERBOSE for readable patterns
pattern = r'''
    \w+           # Username part (word characters)
    @             # Literal @ symbol
    \w+           # Domain name
    \.            # Literal dot
    \w+           # Top-level domain
'''

match = re.search(pattern, text, re.VERBOSE)
if match:
    print(f"Found email: {match.group()}")
    print("✓ re.VERBOSE allows comments and whitespace in patterns")
```

### 🔹 Sub-topic: `g` (Global match)

**Explanation:**
- **⚠️ Not applicable in Python `re` module**
- Python's `re.findall()` and `re.finditer()` already find all matches
- In other languages, `g` flag makes regex find all matches instead of just the first

**What problem it solves:** Finding all matches in a string (in other languages).

**Python behavior:** `re.findall()` and `re.finditer()` already find all matches by default.


```python
# Note: Python's re module doesn't have a 'g' flag
# re.findall() and re.finditer() already find all matches

text = "cat dog cat bird cat"

# Python automatically finds all matches
all_matches = re.findall(r'cat', text)
print(f"All matches: {all_matches}")
print("✓ Python's re.findall() finds all matches by default (no 'g' flag needed)")

# Using finditer for match objects
for match in re.finditer(r'cat', text):
    print(f"  Found 'cat' at position {match.start()}")
```

### 🔹 Sub-topic: `e` (Evaluate replacement)

**Explanation:**
- **⚠️ Not supported in Python's `re` module**
- In some languages, allows evaluation of replacement strings
- Python uses `re.sub()` with a function for dynamic replacements

**What problem it solves:** Dynamic replacement based on matched content (in other languages).

**Python alternative:** Use `re.sub()` with a function: `re.sub(pattern, lambda m: ..., text)`


```python
# Note: Python doesn't have 'e' flag for evaluation
# We use re.sub() with a function instead

text = "Price: 100, Price: 200, Price: 50"

# Python way: Use function for dynamic replacement
def double_price(match):
    price = int(match.group(1))
    return f"Price: {price * 2}"

result = re.sub(r'Price: (\d+)', double_price, text)
print(f"Original: {text}")
print(f"After replacement: {result}")
print("✓ Python uses re.sub() with function instead of 'e' flag")
```

### 🔹 Sub-topic: `U` (Ungreedy pattern)

**Explanation:**
- **⚠️ Not directly supported in Python's `re` module**
- Makes all quantifiers lazy by default
- In Python, use `?` after quantifiers to make them lazy

**What problem it solves:** Making quantifiers lazy by default (in other languages).

**Python alternative:** Add `?` after quantifiers: `*?`, `+?`, `??`, `{n,m}?`


```python
# Note: Python doesn't have 'U' flag for ungreedy
# We use ? after quantifiers to make them lazy

text = 'Say "hello" and "goodbye" to friends'

# Python way: Use ? after quantifier for lazy matching
# *? is lazy (matches as little as possible)
lazy_pattern = r'"(.*?)"'
lazy_matches = re.findall(lazy_pattern, text)
print(f"Lazy matches: {lazy_matches}")

# Compare with greedy (default)
greedy_pattern = r'"(.*)"'
greedy_match = re.search(greedy_pattern, text)
if greedy_match:
    print(f"Greedy match: {greedy_match.group(1)}")
print("✓ Python uses ? after quantifiers for lazy matching (no 'U' flag)")
```

## 3️⃣ Topic-level Real-world Problems (Integration)

### Combined Example: Using Multiple Flags Together


**Problem:** Parse a multiline log file with case-insensitive error matching


```python
# Problem: Extract error messages from multiline log (case-insensitive)
log = """INFO: System started
error: Connection failed
ERROR: Timeout occurred
info: Process completed
Error: Invalid input"""

# Solution: Combine MULTILINE and IGNORECASE flags
# MULTILINE: ^ matches start of each line
# IGNORECASE: matches "error" in any case
pattern = r'^error:\s*(.+)'
flags = re.MULTILINE | re.IGNORECASE

matches = re.findall(pattern, log, flags)
print("Error messages found:")
for i, error in enumerate(matches, 1):
    print(f"  {i}. {error}")

print("\\n✓ Combined flags: MULTILINE | IGNORECASE")
```
