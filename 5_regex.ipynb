```python
import re
```

```python
# LEVEL 1 — Fundamentals
```

```python
# 1. Match a word

# Task: Match the word "python" exactly
# Input: "I love python programming"

# Expected match: python

txt = "I love python programming"

patterns = re.compile(r'python')

for pattern in re.findall(patterns, txt):
    print(pattern)

# re.findall(patterns, txt)
```

```python
# 2. Match any digit

# Task: Extract all digits
# Input: "Order number: 45821"

# Expected: ['4','5','8','2','1']

txt = "Order number: 45821"

patterns = re.compile(r'\d')

re.findall(patterns, txt)
```

```python
# 3. Match a 4-digit year

# Input: "Born in 1998"

# Expected: 1998

txt = "Born in 1998"

patterns = re.compile(r'\d{4}')

for pattern in re.findall(patterns, txt):
    print(pattern)
```

```python
# 4. Match string start

# Task: Check if string starts with "Hello"
# Input: "Hello world"

txt = "Hello world"

patterns = re.compile(r'^Hello')

if re.search(patterns, txt):
    print(True) 
```

```python
# 5. Match string end

# Task: Check if string ends with "done"
# Input: "Task is done"

txt = "Task is done"

patterns = re.compile(r'done$')

if re.search(patterns, txt):
    print(True) 
```

```python
# LEVEL 2 — Character Classes & Quantifiers
```

```python
# 6. Match lowercase letters only

# Input: "abcXYZ"

# Expected: abc

txt = "abcXYZ"

patterns = re.compile(r'[a-z]+')

for pattern in re.findall(patterns, txt):
    print(pattern)
```

```python
# 7. Match a phone number (simple)

# Format: 123-456-7890

txt = '123-456-7890'

patterns = re.compile(r'\d{3}-\d{3}-\d{4}')

if re.search(patterns, txt):
    print(True) 
```

```python
# 8. Match words with exactly 5 letters

# Input: "apple banana grape peach"

# Expected: apple grape peach

txt = "apple banana grape peach"

patterns = re.compile(r'(\b.{5}\b)')

re.findall(patterns, txt)
```

```python
# 9) Extract multiple spaces

# Task: Extract consecutive spaces (length ≥ 2)
# Input: "Too many spaces"
# Expected Output: [' ', ' ']

txt = "Too many spaces"

patterns = re.compile(r'\s')

re.findall(patterns, txt)
```

```python
# 10) Validate optional country code

# Task: Validate phone with optional country code
# Input: "+91-9876543210"
# Expected Output: True

txt = "+91-9876543210"

patterns = re.compile(r'(^\+\d+)')

if re.search(patterns, txt):
    print(True) 
```

```python
# LEVEL 3 — Grouping & Alternation
```

```python
# 11) Extract specific words

# Task: Extract "cat" or "dog"
# Input: "I have a cat and a dog"
# Expected Output: ['cat', 'dog']

txt = "I have a cat and a dog"

patterns = re.compile(r'cat|dog')

re.findall(patterns, txt)
```

```python
# 12) Extract area code

# Task: Extract area code from phone number
# Input: "(416)-555-1234"
# Expected Output: "416"

txt = "(416)-555-1234"

patterns = re.compile(r'(\d{3})')

re.findall(patterns, txt)[0]
```

```python
# 13) Validate specific words (case-insensitive)

# Task: Validate "yes", "no", or "maybe"
# Input: "YES"
# Expected Output: True

txt =  "YES"

patterns = re.compile(r'[a-z]+', re.IGNORECASE)

if re.findall(patterns, txt):
    print(True)
```

```python
# 14) Extract email domain

# Task: Extract domain from email
# Input: "user@gmail.com"
# Expected Output: "gmail.com"

txt = "user@gmail.com"

patterns = re.search(r'(\w+)@(\w+\.\w+)', txt)

patterns.group(2)
```

```python
# 15) Validate file extension

# Task: Validate .jpg, .png, or .pdf
# Input: "report.pdf"
# Expected Output: True

txt = "report.pdf"

pattern = r'(\w+)(\.)(jpg|png|pdf)'

patterns = re.search(pattern, txt)

try:
    if patterns.group(3):
        print(True)
except:
    None
```

```python
# LEVEL 4 — Boundaries & Lookarounds
```

```python
# 16) Extract whole word only

# Task: Extract whole word "car" only
# Input: "car scary cart car"
# Expected Output: ['car', 'car']

txt = "car scary cart car"

patterns = re.compile(r'\bcar\b')

re.findall(patterns, txt)
```

```python
# 17) Validate digit presence

# Task: Check if string contains at least one digit
# Input: "pass123"
# Expected Output: True

txt = "pass123"

if re.search(r'\d+', txt):
    print(True)
```

```python
# 18) Extract number from price

# Task: Extract numeric value
# Input: "$250"
# Expected Output: "250"

txt = "$250"

patterns = re.search(r'(\$)(\d+)', txt)

patterns.group(2)
```

```python
# 19) Extract word before colon

# Task: Extract word before :
# Input: "ERROR: Something failed"
# Expected Output: "ERROR"

txt = "ERROR: Something failed"

patterns = re.compile(r'([\s|\w]+):')

re.findall(patterns, txt)[0]
```

```python
# 20) Extract words not followed by digits

# Task: Extract words not followed by a digit
# Input: "file1 file test2 data"
# Expected Output: ['file', 'data']

txt = "file1 file test2 data"

patterns = re.compile(r'\b[^\s][\D]+\b')

re.findall(patterns, txt)
```

```python
# LEVEL 5 — Validation
```

```python
# 21) Validate email address

# Task: Validate email format
# Input: "a@b.com"
# Expected Output: True

txt = "a@b.com"

if re.search(r'[a-zA-Z0-9-_\.]+\@[a-zA-Z]+\.[a-zA-Z]+', txt):
    print(True)

```

```python
# 22) Validate IPv4 address

# Task: Validate IPv4 format
# Input: "192.168.1.1"
# Expected Output: True

txt = "192.168.1.1"

if re.search(r'\d{3}\.\d{3}\.\d\.\d', txt):
    print(True)

```

```python
# 23) Validate strong password

# Task: Validate strong password
# Input: "Abcdef1!"
# Expected Output: True

txt = "Abcdef1!"

if re.search(r'[a-zA-Z0-9\W]+', txt):
    print(True)
```

```python
# 24) Validate date

# Task: Validate date in YYYY-MM-DD
# Input: "2024-01-30"
# Expected Output: True

txt = "2024-01-30"

if re.search(r'\d{4}-\d{2}-\d{2}', txt):
    print(True)
```

```python
# 25) Validate URL

# Task: Validate HTTP/HTTPS URL
# Input: "https://google.com"
# Expected Output: True

txt = "https://google.com"

if re.search(r'[http|https]+://[\w]+\.[\w]+', txt):
    print(True)
```

```python
# LEVEL 6 — Extraction & Parsing
```

```python
# Task: Extract all HTML tag names
# Input: "<div><p>Hello</p></div>"
# Expected Output: ['div', 'p', 'p', 'div']

lst = []

txt = "<div><p>Hello</p></div>"

patterns = re.compile(r'([<|</])([\w]+)(>)')

for i in re.findall(patterns, txt):
    lst.append(i[1])

lst
```

```python
# 27) Extract numbers in parentheses

# Task: Extract numbers inside parentheses
# Input: "Order(123) ID(456)"
# Expected Output: ['123', '456']
lst = []

txt = "Order(123) ID(456)"

patterns = re.compile(r'(\()([\w]+)(\))')

for i in re.findall(patterns, txt):
    lst.append(i[1])

lst
```

```python
# 28) Extract quoted strings

# Task: Extract text inside double quotes
# Input: He said "hello" and "bye"
# Expected Output: ['hello', 'bye']
lst = []

txt = """
He said "hello" and "bye"
"""

patterns = re.compile(r'(\")([\w]+)(\")')

for i in re.findall(patterns, txt):
    lst.append(i[1])

lst
```

```python
# 29) Extract hashtags

# Task: Extract hashtags without #
# Input: "#python #regex #100DaysOfCode"
# Expected Output: ['python', 'regex', '100DaysOfCode']

txt = "#python #regex #100DaysOfCode"

patterns = re.compile(r'[^\W]+')

re.findall(patterns, txt)
```

```python
# 30) Extract function names

# Task: Extract Python function names
# Input:

# def add(a,b):
#     pass
# def subtract(a,b):
#     pass


# Expected Output: ['add', 'subtract']
lst = []

txt = """
def add(a,b):
    pass
def subtract_bro(a,b):
    pass
"""

patterns = re.compile(r'(def)\s([\w|-]+)')

for i in re.findall(patterns, txt):
    lst.append(i[1])

lst
```

```python
# 31) Validate nested parentheses

# Task: Validate nested parentheses (max depth = 2)
# Input: "(a(b)c)"
# Expected Output: True

txt = "(09awghsj09(bshsk)099csbs0994)"

patterns = re.compile(r'(\()([\w]+)(\()([\w]+)(\))([\w]+)(\))')

if len(re.findall(patterns, txt)) > 0:
    print(True)
```

```python
# 32) Extract key-value pairs

# Task: Extract key-value pairs into dictionary
# Input: "name=John age=30 city=Toronto"
# Expected Output: {'name': 'John', 'age': '30', 'city': 'Toronto'}
dic = {}

txt = "name=John age=30 city=Toronto"

patterns = re.compile(r'([\w|\d]+)=([\w|\d]+)')

for i in re.findall(patterns, txt):
    dic[i[0]] = i[1]
    
dic

```

```python
# 33) Mask credit card number

# Task: Mask all but last 4 digits
# Input: "1234-5678-9012-3456"
# Expected Output: "XXXX-XXXX-XXXX-3456"

txt = "1234-5678-9012-3456"

re.sub(r'([\d]{4})-([\d]{4})-([\d]{4})-([\d]{4})', r'XXXX-XXXX-XXXX-\4', txt)
```

```python
# 34) Extract duplicate words

# Task: Extract consecutive duplicate words
# Input: "this is is a test test"
# Expected Output: ['is', 'test']

txt= "this is is a test test"

patterns = re.compile(r'\b(\w+)\s\1\b')

re.findall(patterns, txt)
```

```python
# 35) Validate balanced quotes

# Task: Validate balanced single or double quotes
# Input: "\"hello\""
# Expected Output: True

txt = "\"hello\""

if re.search(r'^([\"\'])([^\"\']*?)\1$', txt):
    print(True)
```
